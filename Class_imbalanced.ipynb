```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(1) #for reproducibility of result
```

np.random.seed(1) -- means if we genrate random no at any time it will be saME

```python
no_sample = 1000
class_0_ratio = 0.9
no_class_0 = int(no_sample * class_0_ratio)
no_class_1 = no_sample - no_class_0
```

```python
no_class_0
```

```python
no_class_1
```

```python
class_o = {'feature1': np.random.normal(0,1,no_class_0)}
```

```python
class_o = {'feature2': np.random.normal(0,1,no_class_0),'target':[0]*no_class_0}
```

```python
class_0 = pd.DataFrame(class_o)
```

```python
class_0
```

```python
class_1 = pd.DataFrame({'feature1': np.random.normal(3,1,no_class_1)
,'feature2': np.random.normal(0,1,no_class_1),'target':[1]*no_class_1})
```

```python
class_1
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.target.value_counts()
```

```python
df
```

```python
#upsampling
```

```python
df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
df_majority
```

```python
df_minority
```

```python
#replace --- if replace is true , it means data point can be repeated
```

```python
from sklearn.utils import resample
```

```python
df_minority_upsampled = resample(df_minority,replace=True , n_samples=len(df_majority),random_state=1)
```

```python
df_minority_upsampled
```

```python
df_minority_upsampled.shape
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled
```

```python
df_upsampled.target.value_counts()
```

```python
#Down Sampling
```

```python
df_minority
```

```python
df_majority
```

```python
resample(df_majority,replace=False , n_samples=len(df_minority),random_state=1)
```

```python
df_majority_downsampled = resample(df_majority,replace=False , n_samples=len(df_minority),random_state=1)
```

```python
df_downsampled = pd.concat([df_minority,df_majority_downsampled])
```

```python
df_downsampled
```

```python
df_downsampled.target.value_counts()
```
